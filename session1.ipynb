```python
import numpy as np

# Vocabulary
vocabulary = np.array([
    "funny", "great", "bad", "terrible", "emotional",
    "action", "horror", "brutal", "motivational",
    "inspiring", "romantic", "suspensful", "fantasy"
])

print("Vocabulary:", vocabulary)

# Take user preferences
user = []

for word in vocabulary:
    while True:
        user_taste = input(f"Do you like '{word}'? (0/1): ")

        if user_taste in ("0", "1"):
            user.append(int(user_taste))
            break
        else:
            print("Please enter only 0 or 1.")

user = np.array(user)

# Movie database
movies = {
    # Hindi
    "3 idiots": ["funny","great","emotional","motivational","inspiring"],
    "dangal": ["great","emotional","motivational","inspiring"],
    "pk": ["funny","emotional","great"],
    "lagaan": ["great","inspiring","motivational"],
    "taare zameen par": ["emotional","inspiring"],
    "chak de india": ["motivational","inspiring","great"],
    "andhadhun": ["suspensful","great"],
    "drishyam": ["suspensful","great"],
    "kahaani": ["suspensful"],
    "sholay": ["action","great"],
    "kabir singh": ["romantic","emotional"],
    "jab we met": ["romantic","funny"],
    "dilwale dulhania le jayenge": ["romantic","great"],
    "haider": ["emotional","brutal"],
    "gully boy": ["motivational","inspiring"],
    "swades": ["emotional","inspiring"],
    "rang de basanti": ["motivational","inspiring","emotional"],
    "uri": ["action","motivational"],
    "war": ["action"],
    "tumbbad": ["horror","suspensful"],
    "stree": ["funny","horror"],
    "bhool bhulaiyaa": ["funny","horror"],
    "badlapur": ["brutal","suspensful"],
    "don": ["action","suspensful"],
    "raazi": ["suspensful","emotional"],

    # South
    "bahubali": ["action","brutal","great"],
    "bahubali 2": ["action","brutal","great"],
    "kgf": ["action","brutal"],
    "kgf 2": ["action","brutal"],
    "vikram": ["action","brutal","suspensful"],
    "master": ["action"],
    "leo": ["action","brutal"],
    "kantara": ["emotional","great","inspiring"],
    "rrr": ["action","motivational","inspiring"],
    "pushpa": ["action","brutal"],
    "eega": ["fantasy","great"],
    "sita ramam": ["romantic","emotional"],
    "drishyam malayalam": ["suspensful","great"],
    "jallikattu": ["brutal"],
    "asuran": ["brutal","emotional"],
    "jai bhim": ["emotional","inspiring"],
    "super deluxe": ["suspensful"],
    "ratsasan": ["horror","suspensful"],
    "arjun reddy": ["romantic","emotional"],
    "premam": ["romantic","emotional"],
    "96": ["romantic","emotional"],
    "magadheera": ["action","romantic"],
    "ghajini": ["action","emotional"],
    "kaithi": ["action","suspensful"],
    "lucifer": ["action","suspensful"]
}

# Movie names
movie_names = list(movies.keys())

# Feature matrix
x = np.array([
    [1 if word in movies[movie] else 0 for word in vocabulary]
    for movie in movie_names
], dtype=int)

# Calculate scores
scores = np.dot(x, user)

# Best movie
best_index = np.argmax(scores)

print("\nRecommended Movie:", movie_names[best_index])
print("Score:", scores[best_index])

# Optional: Show top 5 recommendations
print("\nTop 5 Recommendations:")
sorted_indices = np.argsort(scores)[::-1]

for idx in sorted_indices[:5]:
    print(f"{movie_names[idx]} --> Score: {scores[idx]}")
```
